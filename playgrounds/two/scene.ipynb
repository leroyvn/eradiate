```python
import logging

import numpy as np
from rich.logging import RichHandler

import eradiate
from eradiate.contexts import KernelContext
from eradiate.two.kernel import DiffuseMaterial, Scene
from eradiate.two.kernel.spectra import InterpolatedSpectrum
from eradiate.units import unit_registry as ureg

logging.basicConfig(
    level="DEBUG",
    format="%(message)s",
    datefmt="[%X]",
    handlers=[RichHandler(rich_tracebacks=True, show_path=False)],
)
logging.getLogger("eradiate.two").setLevel("DEBUG")

eradiate.set_mode("mono")
```

```python
s = InterpolatedSpectrum(
    wavelengths=np.linspace(300, 400, 11),
    values=np.linspace(0, 1, 11),
)
mat = DiffuseMaterial(reflectance=s)

scene = Scene(
    materials={"lambertian": mat},
)
scene.init()
scene
```

```python
import mitsuba as mi

for w in [0.3, 0.35, 0.4] * ureg.um:
    ctx = KernelContext({"w": w})
    scene.update(ctx)
    params = mi.traverse(scene.mi_scene)
    print(f"{w = }")
    print(f"{params['02_bsdf_lambertian.reflectance.value'] = }")
```
